```python
import re
import pandas as pd
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory

# 1. Inisialisasi Sastrawi (Stopwords & Stemming)
stop_factory = StopWordRemoverFactory()
stopwords = stop_factory.get_stop_words()
stem_factory = StemmerFactory()
stemmer = stem_factory.create_stemmer()

# Kumpulan 5 Dokumen (Berita/Pengumuman Fakultas Teknik UNM)
documents = [
    "Pengumuman resmi Fakultas Teknik UNM: Jadwal ujian akhir semester ganjil tahun 2026 telah dirilis. Mahasiswa diharap segera mempersiapkan diri!",
    "Workshop pengembangan perangkat lunak dan web engineering oleh GDG akan diadakan minggu depan di laboratorium komputer. Jangan lupa mendaftar.",
    "Implementasi metode Waterfall dalam rekayasa perangkat lunak selalu terdiri dari analisis, desain, implementasi, pengujian, dan pemeliharaan sistem.",
    "Penggunaan Firebase Realtime Database sangat memudahkan mahasiswa Teknik Komputer angkatan 2024 dalam membangun dan menguji aplikasi mobile.",
    "Mahasiswa diperbolehkan mengajukan form peminjaman fasilitas laboratorium komputer ke UPA TIK untuk keperluan pengerjaan tugas akhir."
]

def preprocess_text(text):
    # a. Case folding
    text = text.lower()

    # b. Cleaning (menghapus angka, tanda baca, karakter khusus)
    text = re.sub(r'[^a-z\s]', ' ', text)

    # c. Tokenisasi
    tokens = text.split()

    # d. Stopwords removal
    tokens_no_stop = [word for word in tokens if word not in stopwords]

    # e. Stemming
    # Stemmer Sastrawi menerima string, jadi kita gabungkan token lalu split kembali
    text_joined = ' '.join(tokens_no_stop)
    text_stemmed = stemmer.stem(text_joined)

    final_tokens = text_stemmed.split()
    return tokens, final_tokens # Mengembalikan token asli (sebelum) dan hasil akhir (sesudah)

# 2 & 3. Terapkan Fungsi dan Tampilkan Perbandingan
results = []
for i, doc in enumerate(documents):
    tokens_before, tokens_after = preprocess_text(doc)

    count_before = len(tokens_before)
    count_after = len(tokens_after)
    reduction = ((count_before - count_after) / count_before) * 100 if count_before > 0 else 0

    results.append({
        "Dokumen": f"Doc {i+1}",
        "Teks Asli": doc,
        "Token Sebelum": tokens_before,
        "Jumlah Sebelum": count_before,
        "Token Sesudah": tokens_after,
        "Jumlah Sesudah": count_after,
        "Pengurangan (%)": round(reduction, 2)
    })

# Menampilkan hasil komprehensif menggunakan pandas DataFrame
df_results = pd.DataFrame(results)

print("--- PERBANDINGAN SEBELUM DAN SESUDAH PREPROCESSING (DOC 1 & 2) ---")
for i in range(2):
    print(f"Dokumen {i+1}:")
    print(f"Asli    : {results[i]['Teks Asli']}")
    print(f"Sebelum : {results[i]['Token Sebelum']}")
    print(f"Sesudah : {results[i]['Token Sesudah']}\n")

# 4. Hitung dan Tampilkan Statistik Token
print("--- STATISTIK PENGURANGAN TOKEN ---")
display(df_results[['Dokumen', 'Jumlah Sebelum', 'Jumlah Sesudah', 'Pengurangan (%)']])
```

Analisis Preprocessing:
Preprocessing menghilangkan noise berupa tanda baca, angka, dan kata sambung (stopwords) yang tidak memiliki nilai pencarian spesifik, serta menyatukan variasi imbuhan menjadi akar kata (stemming). Dampak utamanya terhadap kualitas data sistem Information Retrieval (IR) adalah penurunan dimensi vocabulary secara drastis yang membuat ukuran indeks menjadi jauh lebih kecil dan ringan. Hal ini tidak hanya mempercepat komputasi saat pencarian dokumen teknis atau berita, tetapi juga meningkatkan relevansi (recall) karena sistem akan mencocokkan akar makna kata pencari secara tepat tanpa terganggu oleh perbedaan struktur tata bahasa dokumen aslinya.
